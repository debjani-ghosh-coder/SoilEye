# Crop Recommendation System #

```python
crop=pd.read_csv('CropData.csv')
crop=crop[['N', 'P', 'K', 'temperature', 'humidity', 'ph',
       'rainfall', 'Sand%', 'Clay%', 'Silt%', 'label']]
crop
```

```python
crop.columns
```

```python
crop=crop[['N', 'P', 'K', 'temperature', 'ph', 'rainfall','label']]
```

```python
crop
```

Changing Rainfall column to Moisture!

```python
crop.rename(columns = {'rainfall':'Moisture','N':'Nitrogen','P':'Phosphorous','K':'Potassium','ph':'pH','temperature':'Temperature'}, inplace = True)
```

```python
crop
```

```python
X=crop[['Temperature','Moisture','Nitrogen', 'Phosphorous', 'Potassium','pH']]
```

```python
Y=crop['label']
```

```python
sensor.columns
```

```python
sensor=sensor[['Temperature','Moisture','Nitrogen', 'Phosphorous', 'Potassium','pH']]
sensor
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test=train_test_split(X,Y,test_size=0.2)
```

```python
X_train
```

```python
from sklearn.preprocessing import MinMaxScaler

mms = MinMaxScaler()
X_train_sc= mms.fit_transform(X_train)

X_test_sc = mms.transform(X_test)
```

SVM Classifier

```python
from sklearn import svm
clf = svm.SVC(kernel='linear')
clf.fit(X_train_sc, Y_train)
clf.score(X_test_sc,Y_test)
```

DTC Classifier

```python
from sklearn.tree import DecisionTreeClassifier

dtc= DecisionTreeClassifier(random_state=42).fit(X_train_sc, Y_train)
dtc.score(X_test_sc,Y_test)
```

Gradient Boosting Classifier

```python
from sklearn.ensemble import GradientBoostingClassifier
grad = GradientBoostingClassifier()
grad.fit(X_train_sc, Y_train)

print('Gradient Boosting accuracy : {}'.format(grad.score(X_test_sc,Y_test)))
```

Random Forest Classifier

```python
from sklearn.ensemble import RandomForestClassifier
rfc = RandomForestClassifier(max_depth=14,n_estimators=1000)
rfc.fit(X_train_sc, Y_train)
rfc.score(X_test_sc, Y_test)
```

```python

print('RF Accuracy on training set: ',rfc.score(X_train_sc, Y_train))
print('RF Accuracy on test set: ',rfc.score(X_test_sc, Y_test))
```
